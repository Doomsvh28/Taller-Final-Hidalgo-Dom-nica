```python
#Doménica Hidalgo
import random
class usuario:
  def __init__(self, nombre, tipo, saldo):
    self.saldo = float(saldo)
    self.nombre = nombre
    self.tipo = tipo
    if tipo == "1":
      "Normal"
    else:
       "Premium"
    self.historial = []



  def recargar_saldo(self, monto):
    if monto > 0:
      self.saldo += monto
      print(f"Recarga realizada! Saldo: {self.saldo}")
    else:
      print("Error")

  def realizar_viaje(self, km):
    costo = km * (0.40 if self.tipo == "Premium" else 0.50)
    if self.saldo > costo:
        self.saldo -= costo
        self.historial.append(km)

        viaje= random.randint(100,999)
        print(f"Viaje {viaje} realizado! Costo: ${costo}. Saldo: {self.saldo}")
        return costo
    else:
      print("Saldo insuficiente")
      return 0

  def ver_historial(self):
    print("=== HSITORIAL ===")
    for i, km in enumerate(self.historial, 1):
      print(f"Viaje {i}: {km} km")

  def ver_perfil(self, detallado=True):
    print("\n=== PERFIL ACTUAL ===")
    print(f"Nombre: {self.nombre}")
    print(f"Saldo: ${self.saldo}")
    if detallado:
      print(f"Tipo: {self.tipo}")
      print(f"Viajes realizados: {len(self.historial)}")

  def calcular_gasto_total(self):
    des = 0.40 if self.tipo =="Premium" else 0.50
    return sum(self.historial)*des

class premium(usuario):
  def __init__(self, nombre, saldo):
    super().__init__(nombre, saldo)
    self.tipo = "Premium"
  def calcular_precio(self,km):
    return (km*0.50)*0.50

  def realizar_viaje(self, km):
    if km <= 0:
      print("Error")
      return 0
    return super().realizar_viaje(km)



nombre= str(input("Nombre: "))
tipo = int(input("Tipop (1 o 2): "))
saldo = float(input("Saldo inicial: "))

if tipo == "2":
  persona = premium(nombre, saldo)
else:
  persona= usuario(nombre,"1", saldo)

while True:
  persona.ver_perfil(detallado=True)
  print("==MENU==")
  print("1. Recargar saldo")
  print("2. Realizar viaje")
  print("3. Ver historial")
  print("4. Ver perfil")
  print("5. Ver gasto total")
  print("6. Salir")

  opcion = int(input("Opcion: "))
  if opcion == 1:
    m = float(input("Monto a recargar: "))
    persona.recargar_saldo(m)
  elif opcion ==2:
    dist = float(input("Kilómetros: "))
    persona.realizar_viaje(dist)
  elif opcion ==3:
    persona.ver_historial()
  elif opcion == 4:
    persona.ver_perfil(detallado=True)
  elif opcion == 5:
    total= persona.calcular_gasto_total()
    print(f"Suma total de costos de viajes: ${total}")

  elif opcion == 6:
    print("Gracias por usar el servicio")
    break
  else:
    print("Error")










```
